### Крок 5: Індексація за допомогою loc та iloc

```python
row_iloc = df_sales.iloc[2]
print(row_iloc)
```

```python
row_loc = df_sales.loc[2]
print(row_loc)
```

```python
value = df_sales.loc[5, 'money']
print(value)
```

Різниця між `loc` та `iloc` стає помітною, коли індекси не збігаються з порядковими номерами (наприклад, після сортування — див. крок 7).

### Крок 6: Фільтрація даних за умовою

```python
mask_high = df_sales['money'] > 35
high_sales = df_sales[mask_high]
print(high_sales.head())
```

### Крок 7: Сортування даних

```python
sorted_sales = df_sales.sort_values('money', ascending=False)
print(sorted_sales.head())
```

```python
print("iloc[0] — перший рядок у відсортованому порядку:")
print(sorted_sales.iloc[0])
print()
print("loc[0] — рядок з оригінальною міткою індексу 0:")
print(sorted_sales.loc[0])
```

Індекси рядків після сортування залишились від оригінальної таблиці, тому `iloc[0]` і `loc[0]` повертають **різні** рядки.

### Крок 8: Збереження результатів

```python
final = df_sales[df_sales['money'] > 35].sort_values('money', ascending=False)
final.to_csv('/content/drive/MyDrive/filtered_sales.csv', index=False)
print("Файл збережено!")
```
