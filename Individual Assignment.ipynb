### ★ Додаткове завдання: функція підрахунку записів за категоріальною колонкою

```python
def count_by_category(df, column):
    """Повертає DataFrame з кількістю записів у кожній групі колонки `column`,
    відсортований за кількістю за спаданням."""
    counts = df[column].value_counts().reset_index()   # кількість записів у групах
    counts.columns = [column, 'count']
    return counts.sort_values('count', ascending=False)


coffee_counts = count_by_category(df_sales, 'coffee_name')
print(coffee_counts)

coffee_counts.to_csv('/content/drive/MyDrive/coffee_counts.csv', index=False)
print("Файл збережено!")
```

## ПИТАННЯ ДЛЯ САМОПЕРЕВІРКИ

1. **DataFrame і Series.** Series — одновимірний масив із мітками (стовпець таблиці). DataFrame — двовимірна таблиця з кількох Series зі спільним індексом.
2. **DataFrame зі словника:** `pd.DataFrame({'a': [...], 'b': [...]})`; ключі — назви колонок. Усі списки мають бути **однакової довжини**, інакше буде помилка.
3. **Початок і кінець таблиці:** `head()` та `tail()`.
4. **info() та describe():** `info()` — технічна інформація (колонки, кількість непорожніх значень, типи, пам'ять); `describe()` — статистики (кількість, середнє, std, min, квартилі, max) для числових стовпців.
5. **`df['price'] > 35`** повертає Series з булевими значеннями (маска). Фільтрація: `df[df['price'] > 35]`.
6. **loc та iloc:** `loc` — за мітками індексу, `iloc` — за позицією (з 0). Після сортування `df.iloc[0]` — перший рядок у новому порядку, а `df.loc[0]` — рядок із міткою 0.
7. **Конкретне значення:** за позицією — `df.iloc[2, df.columns.get_loc('score')]`; за міткою — `df.loc[1, 'score']`.
8. **Сортування:** `sort_values()`; за двома колонками — `df.sort_values(['col1', 'col2'])`.
9. **sort_values()** за замовчуванням повертає новий об'єкт і не змінює оригінал; щоб змінити — `inplace=True` (або присвоїти результат тій самій змінній).
10. **Збереження без індексу:** `df.to_csv('file.csv', index=False)`.
11. **Ланцюжок:** `df[df['money'] > 35].sort_values('money', ascending=False)`.
12. **`and` замість `&`:** виникне `ValueError` («truth value of a Series is ambiguous»), бо `and` намагається перетворити весь Series на одне True/False, а `&` працює поелементно.
